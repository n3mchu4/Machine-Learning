## Bài 1. Tạo Tensor cơ bản

```python
import torch

a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.int32
)
print(a)

print("Shape:", a.shape)
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

**Nhận xét:**
- Tensor a có kích thước (shape) torch.Size([2, 3]), tức 2 hàng × 3 cột.
- Số chiều (ndim) = 2 (ma trận 2 chiều).
- Kiểu dữ liệu (dtype) là torch.int32 vì được khai báo tường minh trong lúc tạo Tensor.

## Bài 2. Tạo các Tensor thường dùng

```python
import torch
torch.manual_seed(42)

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print("a (rand):")
print(a)
print("b (randn):")
print(b)
print("c (zeros):")
print(c)
print("d (ones):")
print(d)
print("e (eye):")
print(e)

# Yeu cau bo sung: Tensor 5x5 toan gia tri 10
f = torch.full((5, 5), 10)
print("f (full 5x5 = 10):")
print(f)
```

**Nhận xét:**
- torch.rand() sinh số ngẫu nhiên theo phân phối **đều (uniform)** trong khoảng [0, 1), nên tất cả giá trị của a đều dương và nhỏ hơn 1.
- torch.randn() sinh số ngẫu nhiên theo phân phối **chuẩn (normal)** với trung bình 0, độ lệch chuẩn 1, nên b có cả giá trị âm và dương, và có thể lớn hơn 1 hoặc nhỏ hơn -1.
- Tensor 5×5 toàn giá trị 10 được tạo bằng torch.full((5, 5), 10).

## Bài 3. Kiểm tra thông tin Tensor

## Bài 4. Slicing và Indexing

```python
import torch
torch.manual_seed(42)

a = torch.randn(3, 4, 5)

print("Original tensor:")
print(a)

print("\nTensor a[1]:")
print(a[1])
print("Shape a[1]:", a[1].shape)

print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
print("Shape a[1:, 2:4]:", a[1:, 2:4].shape)

# Yeu cau bo sung: lay hai phan tu cuoi tren chieu thu ba
print("\nHai phan tu cuoi tren chieu thu ba (a[:, :, -2:]):")
print(a[:, :, -2:])
print("Shape:", a[:, :, -2:].shape)
```

**Nhận xét:**
- a[1] loại bỏ chiều đầu tiên nên shape là [4, 5].
- a[1:, 2:4] giữ nguyên số chiều (3 chiều) nhưng thu hẹp kích thước từng chiều: chiều 0 còn 2 phần tử (từ index 1 đến hết), chiều 1 còn 2 phần tử (index 2,3) → shape [2, 2, 5]
- Để lấy 2 phần tử cuối trên chiều thứ ba (chiều index=2), dùng slicing a[:, :, -2:], kết quả có shape [3, 4, 2]

## Bài 5. Reshape và Flatten

```python
import torch
torch.manual_seed(42)

a = torch.randn(3, 4)

print("Original:")
print(a)

print("\nFlatten:")
print(a.ravel())

print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))

print("\nSo phan tu truoc reshape:", a.numel())
print("So phan tu sau reshape:", a.reshape(3, 2, 2).numel())

print("\nReshape to 2 x 6:")
print(a.reshape(2, 6))
```

**Nhận xét:**
- Tensor gốc có 12 phần tử (3×4). Sau khi ravel() (flatten) hoặc reshape(3, 2, 2), số phần tử vẫn giữ nguyên là 12 — reshape chỉ thay đổi cách sắp xếp, không thay đổi dữ liệu hay tổng số phần tử.
- Reshape thành (2, 6) cũng hợp lệ vì 2×6 = 12.

## Bài 6. Phép toán và nhân ma trận

```python
import torch
torch.manual_seed(42)

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication (a @ b.T):")
print(a @ b.T)
print("Shape a @ b.T:", (a @ b.T).shape)

print("\na * b (element-wise):")
print(a * b)
```

**Nhận xét:**
- a * b là phép **nhân theo từng phần tử (element-wise)**: kết quả có cùng shape (2, 3) với a và b, mỗi phần tử là tích của hai phần tử tương ứng.
- a @ b.T là **phép nhân ma trận (matrix multiplication)**:a có shape (2,3), b.T có shape (3,2), nên kết quả có shape (2, 2) — mỗi phần tử là tích vô hướng giữa một hàng của a và một hàng của b.

## Bài 7. Các hàm thống kê trên Tensor

```python
import torch
torch.manual_seed(42)

a = torch.randn(3, 4)
print(a)

print("\nMean (dim=0):")
print(torch.mean(a, dim=0))

print("\nStandard deviation (dim=0):")
print(torch.std(a, dim=0))

print("\nCumulative sum (dim=0):")
print(torch.cumsum(a, dim=0))

print("\n--- dim=1 ---")
print("\nMean (dim=1):")
print(torch.mean(a, dim=1))

print("\nStandard deviation (dim=1):")
print(torch.std(a, dim=1))

print("\nCumulative sum (dim=1):")
print(torch.cumsum(a, dim=1))
```

**Nhận xét:**
- Với dim=0: các phép toán mean, std, cumsum được tính **theo từng cột** thu gọn theo chiều hàng, kết quả có 4 giá trị ứng với 4 cột.
- Với dim=1: các phép toán được tính **theo từng hàng** thu gọn theo chiều cột, kết quả có 3 giá trị ứng với 3 hàng.
- Nói chung, tham số dim cho biết chiều sẽ bị "co lại" reduce khi tính toán; các chiều còn lại được giữ nguyên.

## Bài 8. Autograd cơ bản

```python
import torch

x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

print("\nDao ham thu cong: dy/dx = 2x = 2 * 3.6 =", 2 * 3.6)

# Doi ham
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()
print("\ny2 = 3x^2 + 2x + 1")
print("x2.grad =", x2.grad)
print("Dao ham thu cong: dy/dx = 6x + 2 =", 6 * 3.6 + 2)
```

**Nhận xét:**
- Với `y = x²`, đạo hàm `dy/dx = 2x`. Tại `x = 3.6`: `dy/dx = 7.2`, khớp hoàn toàn với `x.grad` do autograd tính tự động.
- Với `y = 3x² + 2x + 1`, đạo hàm `dy/dx = 6x + 2`. Tại `x = 3.6`: `dy/dx = 23.6`, cũng khớp với `x.grad`.
- Điều này cho thấy autograd của PyTorch tính đạo hàm chính xác bằng phương pháp vi phân tự động (automatic differentiation), không phải xấp xỉ số.

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

np.random.seed(0)
torch.manual_seed(0)

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
print("\nHe so ky vong: [1, 2, 3]")
print("MSE cuoi cung:", mse.item())
```

**Nhận xét:**
- Ba hệ số học được xấp xỉ `[1.0017, 2.0058, 2.8337]`, rất gần với hệ số thực của đa thức gốc `[1, 2, 3]`. Sai lệch nhỏ là do chỉ chạy 1000 epoch và dữ liệu có nhiễu ngẫu nhiên trong quá trình sinh (`X`), MSE cuối đã giảm về mức thấp (~0.018).
- **`optimizer.zero_grad()`**: đưa gradient tích lũy từ vòng lặp trước về 0, tránh việc gradient bị cộng dồn qua các epoch (vì PyTorch mặc định cộng dồn gradient).
- **`mse.backward()`**: thực hiện lan truyền ngược (backpropagation), tính gradient của hàm mất mát `mse` theo tham số `w` và lưu vào `w.grad`.
- **`optimizer.step()`**: cập nhật giá trị tham số `w` dựa trên gradient vừa tính, theo quy tắc của thuật toán tối ưu (ở đây là NAdam).

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch

random.seed(0)
torch.manual_seed(0)

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("\nKiem tra:")
print("A+B =", (A+B).item(), " (kỳ vọng 9)")
print("C-D =", (C-D).item(), " (kỳ vọng 1)")
print("A+C =", (A+C).item(), " (kỳ vọng 8)")
print("B-D =", (B-D).item(), " (kỳ vọng 2)")
```

**Nhận xét:**
- Nghiệm tìm được thỏa mãn rất tốt cả 4 phương trình: A+B ≈ 9, C−D ≈ 1, A+C ≈ 8, B−D ≈ 2 (sai số chỉ khoảng 10⁻⁵).
- Bài toán có **4 ẩn số (A, B, C, D)** nhưng chỉ có **4 phương trình**, tuy nhiên các phương trình này không đủ để xác định duy nhất từng biến theo kiểu tuyến tính độc lập hoàn toàn — thực chất hệ có hạng (rank) nhỏ hơn 4 do các phương trình có thể phụ thuộc tuyến tính lẫn nhau một phần, dẫn tới **vô số nghiệm** thỏa mãn hệ (chỉ các *hiệu*/*tổng* giữa các biến bị ràng buộc, còn có thể cộng thêm một hằng số vào cả 4 biến mà vẫn giữ nguyên các hiệu số). Vì vậy, giá trị cụ thể của A, B, C, D phụ thuộc vào điểm khởi tạo ngẫu nhiên ban đầu.

## Bài 11. Bài tập áp dụng

```python
import numpy as np
import torch

torch.manual_seed(0)
np.random.seed(0)

print("=== Phan 1: Tao Tensor X va reshape ===")
X = torch.arange(24)
print("X:", X)
print("Reshape 4x6:")
print(X.reshape(4, 6))
print("Reshape 2x12:")
print(X.reshape(2, 12))
print("Reshape 2x3x4:")
print(X.reshape(2, 3, 4))

print("\n=== Phan 2: Cac phep toan tren A, B (3x4) ===")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A:")
print(A)
print("B:")
print(B)
print("A + B:")
print(A + B)
print("A * B:")
print(A * B)
print("A / B:")
print(A / B)
print("Trung binh theo cot (dim=0):")
print(torch.mean(A, dim=0))
print("Do lech chuan theo cot (dim=0):")
print(torch.std(A, dim=0))

print("\n=== Phan 3: Autograd voi y = 2x^2 + 5x + 3 ===")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x ** 2 + 5 * x + 3
y.backward()
print("y tai x=2:", y.item())
print("dy/dx (autograd):", x.grad.item())
print("dy/dx (thu cong) = 4x + 5 =", 4 * 2 + 5)

print("\n=== Phan 4: Gradient Descent cho y = 2x^2 - 3x + 5 ===")
polynomial = np.poly1d([2, -3, 5])
N = 20
Xd = np.random.randn(N, 1) * 5
Yd = polynomial(Xd)

XX = np.hstack([
    Xd * Xd,
    Xd,
    np.ones_like(Xd)
])

w = torch.randn(3, 1, requires_grad=True)
xd = torch.tensor(XX, dtype=torch.float32)
yd = torch.tensor(Yd, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("He so ban dau:")
print(w)

for epoch in range(5000):
    optimizer.zero_grad()
    y_pred = xd @ w
    mse = torch.mean(torch.square(yd - y_pred))
    mse.backward()
    optimizer.step()

print("He so hoc duoc:")
print(w)
print("He so ky vong: [2, -3, 5]")
print("MSE cuoi cung:", mse.item())
```

**Nhận xét:**
- **Reshape:** Tensor `X` có 24 phần tử được reshape thành công sang các dạng `4×6`, `2×12`, `2×3×4` vì tích các chiều luôn bằng 24.
- **Phép toán trên A, B (3×4):** `A+B`, `A*B` (element-wise), `A/B` (chia từng phần tử) đều cho kết quả cùng shape `(3,4)`. `mean`/`std` theo `dim=0` cho 4 giá trị tương ứng 4 cột.
- **Autograd cho `y = 2x² + 5x + 3`:** đạo hàm giải tích là `dy/dx = 4x + 5`; tại `x=2`, giá trị này là `13`, khớp hoàn toàn với gradient tính bằng autograd (`x.grad = 13`).
- **Huấn luyện `y = 2x² − 3x + 5`:** với 1000 epoch ban đầu mô hình chưa hội tụ tốt (MSE còn khá lớn); sau khi tăng lên **5000 epoch**, các hệ số học được hội tụ chính xác về `[2.0000, -3.0000, 5.0000]`, khớp với hệ số kỳ vọng, và MSE giảm gần về 0. Điều này cho thấy tầm quan trọng của số epoch (đủ vòng lặp) đối với sự hội tụ của Gradient Descent.
